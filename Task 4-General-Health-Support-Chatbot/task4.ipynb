```python
from huggingface_hub import InferenceClient

HF_TOKEN = "YOUR_HF_TOKEN"

# Initialize the client
client = InferenceClient(model="mistralai/Mistral-7B-Instruct-v0.2", token=HF_TOKEN)

print("Chatbot Client Ready!")
```

```python
# SYSTEM_PROMPT defines the persona and safety constraints
SYSTEM_PROMPT = """
You are a helpful, empathetic, and clear General Health Assistant. 
Your goal is to provide informational answers to health queries.

SAFETY GUIDELINES:
1. You are NOT a doctor. Always include a disclaimer that the user should consult a medical professional.
2. If a query describes an emergency (e.g., chest pain, difficulty breathing), tell the user to call emergency services immediately.
3. Do not prescribe specific dosages for medications.
4. Use friendly, simple language.
"""

def get_health_response(user_query):
    # Combine the system instructions with the user's question
    prompt = f"<s>[INST] {SYSTEM_PROMPT} \nUser Question: {user_query} [/INST]</s>"
    
    response = client.text_generation(prompt, max_new_tokens=500)
    return response

print("Prompt Engineering logic established.")
```

```python
def get_health_response(user_query):
    # We use chat_completion which is the standard for Instruct models
    response = client.chat_completion(
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": user_query},
        ],
        max_tokens=500,
    )
    
    # Extract just the text content from the response object
    return response.choices[0].message.content

print("Updated function to use chat_completion.")
```

```python
print("Type 'quit' to exit the chat.")

# Ensure HF_TOKEN is defined from your setup cell
if 'HF_TOKEN' in locals() and HF_TOKEN != "YOUR_TOKEN_HERE":
    try:
        while True:
            user_input = input("You: ")
            if user_input.lower() == 'quit':
                print("Chat ended. Have a great day!")
                break
                
            response = get_health_response(user_input)
            print(f"\nAssistant: {response}\n")
    except EOFError:
        print("\nChat session ended.")
else:
    print("Cannot start interactive chat without a valid token. Please configure HF_TOKEN in the setup cell first.")
```
