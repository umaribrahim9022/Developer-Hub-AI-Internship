```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = sns.load_dataset('iris')

# 1. Inspect the data
print("Dataset Shape")
print(df.shape)

print("\n Column Names")
print(df.columns)

print("\n First 5 Rows")
print(df.head())
```

```python
print("\nInfo")
df.info()

print("\nDescriptive Statistics")
print(df.describe())
```

```python
plt.figure(figsize=(10, 6))
sns.pairplot(df, hue='species', palette='husl', markers=["o", "s", "D"])
plt.suptitle("Pairwise Scatter Plot of Iris Features", y=1.02)
plt.show()
```

```python
df.hist(edgecolor='black', linewidth=1.2, figsize=(12, 8))
plt.suptitle("Distribution of Features (Histograms)")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
sns.boxplot(x='species', y='sepal_length', data=df)
plt.title("Sepal Length by Species")

plt.subplot(1, 2, 2)
sns.boxplot(x='species', y='petal_width', data=df)
plt.title("Petal Width by Species")

plt.tight_layout()
plt.show()
```
